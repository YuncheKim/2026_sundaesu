```python
def dfs(graph_n, v, visited):
    visited.add(v)
    print(v, end=' ')
    for neighbor in graph_n.get(v, []):
        if neighbor not in visited:
            dfs(graph_n, neighbor, visited)
```

```python
def bfs(graph_n, v, visited):
    queue = [v]
    visited.add(v)
    while queue:
        u = queue.pop(0)
        print(u, end=' ')
        for neighbor in graph_n.get(u, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
```

```python
print("DFS탐색 순서:")
dfs(graph_n, '김윤채', set())
```

```python
print("BFS탐색 순서:")
bfs(graph_n, '김윤채', set())
```

```python
def gamyeom(graph_n, start, max_level=2):
  visited = {start}
  parent = {start: None}
  queue = [(start, 0)]
  print(f"감염자: {start}")
  while queue:
        current, level = queue.pop(0)
        if level == max_level:
            continue
        for neighbor in graph_n.get(current, []):
            if neighbor not in visited:
                visited.add(neighbor)
                parent[neighbor] = current
                queue.append((neighbor, level + 1))
                path, p = [], neighbor
                while p is not None:
                    path.append(p)
                    p = parent[p]
                print(f"[{level + 1}단계] {neighbor}  경로: {' -> '.join(path[::-1])}")

gamyeom(graph_n, '홍길동')
```

```python
import random

def make_graph(n, depth, seed):
    random.seed(seed)                                  # 씨드 고정 → 매번 같은 그래프
    names = [f"P{i}" for i in range(n)]
    graph_n = {name: [] for name in names}

    # 1) 각 노드의 단계(P0까지 거리) 정하기
    level = {}
    for i in range(depth + 1):                         # P0~P{depth}는 0~depth단계로 고정
        level[names[i]] = i                            #  → 가장 먼 노드가 정확히 depth
    for name in names[depth + 1:]:                     # 나머지는 1~depth단계 중 무작위
        level[name] = random.randint(1, depth)

    by_level = {}                                      # 단계별 노드 목록
    for name in names:
        by_level.setdefault(level[name], []).append(name)

    def connect(a, b):                                 # 무방향, 중복·자기 연결 방지
        if a != b and b not in graph_n[a]:
            graph_n[a].append(b)
            graph_n[b].append(a)

    # 2) 뼈대: P0 - P1 - ... - P{depth} 한 줄로 연결
    for i in range(1, depth + 1):
        connect(names[i - 1], names[i])

    # 3) 나머지 노드는 바로 앞 단계 노드 하나와 연결 → 모두 P0과 이어짐
    for name in names[depth + 1:]:
        connect(name, random.choice(by_level[level[name] - 1]))

    # 4) 추가 간선: 같은 단계나 바로 옆 단계끼리만 → 거리가 줄어들지 않음
    for _ in range(n // 3):
        a = random.choice(names)
        lv = level[a] + random.choice([-1, 0, 1])
        if lv in by_level:
            connect(a, random.choice(by_level[lv]))

    return graph_n


def max_depth(graph_n, start):                         # 검증용: 가장 먼 노드까지 거리
    dist = {start: 0}
    queue = [start]
    while queue:
        u = queue.pop(0)
        for v in graph_n[u]:
            if v not in dist:
                dist[v] = dist[u] + 1
                queue.append(v)
    return max(dist.values())


# 테스트
for n, d, s in [(100, 10, 1), (1000, 50, 2)]:
    g = make_graph(n, d, seed=s)
    print(f"\n===== 노드 {len(g)}개, max depth {max_depth(g, 'P0')} =====")
    gamyeom(g, 'P0')
```
